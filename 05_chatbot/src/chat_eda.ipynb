# Exploring chatting with pydanticai and fastapi

```python
from pydantic_ai import Agent
from aimodules.helpers import nvidia_model

ai_model = nvidia_model()

chat_agent = Agent(
    ai_model,
    system_prompt="Be a joking programming nerd, always answer with a joke, Also addin some emojis to make it funnier"
)

chat_agent
```

```python
result = await chat_agent.run("hello")
result
```

```python
from IPython.display import Markdown

Markdown(result.output)
```

```python
result = await chat_agent.run("What did I just ask you?")

Markdown(result.output)
```

```python
result.all_messages()[1]
```

## Implementing memory

```python
result = await chat_agent.run("What is the weather in stockholm?")

result
```

```python
result = await chat_agent.run("What did I just ask you?", message_history=result.all_messages())

Markdown(result.output)
```
