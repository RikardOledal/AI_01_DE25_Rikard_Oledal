## Setup AI model

```python
from pydantic_ai import Agent, ModelSettings
from pydantic_ai.models.openai import OpenAIChatModel
from pydantic_ai.providers.openai import OpenAIProvider
from IPython.display import Markdown
import os
from dotenv import load_dotenv

load_dotenv()

AI_model = OpenAIChatModel(
    model_name="nvidia/nemotron-3.5-lightning-30b-a3b",
    provider=OpenAIProvider(
        base_url = "https://integrate.api.nvidia.com/v1",
        api_key = os.environ["NVIDIA_API"]
        )
    )

AI_model
```

## Setup AI Agent

### Setup Temperature

```python
demo_agent = Agent(
    model=AI_model,
    system_prompt="Answer with max 3 sentence. Start with Once upon... ",
    model_settings=ModelSettings(temperature=0)
    )

prompt = "How do you cook an egg?"

result = await demo_agent.run(prompt)
print(result.output)
```

```python
result = await demo_agent.run(prompt)
print(result.output)
```

### Temperature change
- When you expand temperature it is getting more creative

```python
demo_agent.model_settings = ModelSettings(temperature=1)
result = await demo_agent.run(prompt)
print(result.output)
```
